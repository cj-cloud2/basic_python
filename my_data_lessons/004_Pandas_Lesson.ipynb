# 📌 Topic 2 — Loading Data with `read_csv()`

We use `pd.read_csv()` to load a CSV file into a **DataFrame** — pandas' core data structure (think: a spreadsheet in Python).

Two important parameters:
- `index_col='Symbol'` → use the Symbol column as the row label instead of 0, 1, 2...

# 📌 Topic 3 — Inspecting Data

Before doing any analysis, always take a quick look at your data.

- `.shape` → (rows, columns) — how big is it?
- `.size` → total number of cells (rows × columns)
- `.describe()` → count, mean, min, max, quartiles for every numeric column

```python
# Transpose describe() — sometimes easier to read when there are many columns
df.describe().transpose()
```

```python
# Transpose the whole DataFrame — rows and columns swap
df.transpose()
```

# 📌 Topic 4 — What is an Index?

An **index** is the set of row labels in a DataFrame.

- By default pandas uses numbers: 0, 1, 2, 3...
- You can use any column as the index (like Symbol, Name, Date)
- The index is used to **look up rows** quickly
- Indexes don't have to be unique or sorted (but often they are)

Our `df` uses **Symbol** as the index — let's see it clearly.

```python
# See the index
print('Index:', df.index.tolist())
print()
print('Column index:', df.columns.tolist())
```

```python
# Compare: load WITHOUT an index_col — default numeric index
df_default = pd.read_csv('data/stocks.csv')
print('Default numeric index:')
df_default.head(3)
```

# 📌 Topic 5 — Selecting Columns

Two ways to select columns:
- `df['column']` → returns one column (as a Series)
- `df[['col1', 'col2']]` → returns multiple columns (as a DataFrame)

Note: the order inside the list is the order you get back — you can reorder columns this way!

```python
# Select a single column — returns a Series
df['Sector']
```

```python
# Select multiple columns — returns a DataFrame
df[['Name', 'Price']]
```

```python
# You control the order — swap the columns
df[['Price', 'Name']]
```

# 📌 Topic 6 — Selecting Rows — `.loc[]`, `.iloc[]` & Dual `[]` Behaviour

**The dual behaviour of `df[]`** — this confuses beginners:
- `df['Name']` → looks for a **column** called Name
- `df['a':'c']` → the `:` tells pandas you mean **rows**

The clean way to select rows is always `.loc[]` or `.iloc[]`:
- `.loc['label']` → select by **row label** (the index value)
- `.iloc[3]` → select by **row position** (integer, starts at 0)

```python
# ⚠️  Dual behaviour — colon means ROWS, no colon means COLUMN
# This selects ALL rows (row slice with no start/end)
df[:]   # all rows
```

```python
# Select a row by its INDEX LABEL using .loc[]
df.loc['AAPL']
```

```python
# Select multiple rows by label
df.loc[['AAPL', 'MSFT', 'GOOGL']]
```

```python
# Select a row by its POSITION using .iloc[] (0 = first row)
df.iloc[0]   # first row
```

```python
# .iloc with a range — rows 0, 1, 2 (stop is exclusive)
df.iloc[0:3]
```

```python
# .loc can also select all rows of a specific column
df.loc[:, 'Price']   # all rows, Price column
```

```python
# .iloc — all rows, 3rd column (index starts at 0)
df.iloc[:, 2]
```

# 📌 Topic 7 — Accessing Individual Cells

Once you know `.loc[]` and `.iloc[]`, accessing a single cell is simple — just give **both** a row and a column.

```python
# Access one cell by label — row=AAPL, column=Price
df.loc['AAPL', 'Price']
```

```python
# Access one cell by position — row 0, column 2
df.iloc[0, 2]
```

```python
# Both give the same result for Apple's Price
print('By label:', df.loc['AAPL', 'Price'])
print('By position:', df.iloc[0, 2])
```

# 📌 Topic 8 — Adding Columns

Adding a new column is as simple as assigning to it:
- `df['new_col'] = value` → if the column exists, it **overwrites** it
- The value can be a constant, a list, or a calculation from other columns

```python
# Add a column with a constant value
df['Country'] = 'USA'
df.head()
```

```python
# Add a column calculated from other columns
# Volume in millions = Volume / 1,000,000
df['Vol_M'] = df['Volume'] / 1_000_000
df.head()
```

```python
# ⚠️  Assigning to an existing column OVERWRITES it — no warning!
df['Country'] = 'United States'
df.head()
```

# 📌 Topic 9 — Inserting Columns at a Specific Position

`.insert(loc, column, value)` lets you place a column exactly where you want it.

- `loc` = position number (0 = first, 1 = second...)
- `column` = name of the new column
- `value` = data to put in it

⚠️ Inserting a column name that **already exists** raises an error. Use `drop`/`pop`/`del` first.

```python
# Check column order before insert
print('Before:', df.columns.tolist())
```

```python
# Insert a 'Market' column at position 1 (second column)
df.insert(loc=1, column='Market', value='NYSE')
df.head()
```

```python
# Check column order after insert
print('After:', df.columns.tolist())
```

# 📌 Topic 10 — Removing Rows & Columns

Three tools for removing columns:
| Tool | Removes | Returns | Modifies original? |
|------|---------|---------|--------------------|
| `drop()` | rows or columns | new DataFrame | ❌ No |
| `pop()` | columns only | the removed column | ✅ Yes |
| `del` | columns only | nothing | ✅ Yes |

**Rule of thumb:** Use `drop()` for rows. Use `pop()` for columns (you get the data back). Use `del` only when you don't need the data.

```python
# drop() a ROW — original is UNCHANGED, returns a new DataFrame
result = df.drop('GE')
print('Rows in result:', len(result))
print('Rows in original df:', len(df))   # still 16
```

```python
# drop() multiple rows
df.drop(['GE', 'CAT'])
```

```python
# drop() a COLUMN — must pass axis=1 or axis='columns'
df.drop('Market', axis=1).head()   # original unchanged
```

```python
# pop() removes a column FROM the DataFrame and returns it
removed = df.pop('Country')
print('Removed column values:')
print(removed.head())
print()
print('Columns left:', df.columns.tolist())
```

```python
# del removes a column — returns nothing, very permanent
del df['Market']
print('Columns after del:', df.columns.tolist())
```

# 📌 Topic 11 — Creating a DataFrame from Scratch

You can build a DataFrame directly in code — no CSV needed.
Common ways:
- From a numpy array
- From a dictionary

Useful for testing, teaching, and building small lookup tables.

```python
# Create a 4x4 DataFrame from a numpy array
d    = np.arange(16).reshape(4, 4)   # numbers 0-15 in a 4x4 grid
ndx  = ['a', 'b', 'c', 'd']          # row labels
cols = ['one', 'two', 'three', 'four']  # column labels

df_small = pd.DataFrame(data=d, index=ndx, columns=cols)
df_small
```

```python
# Create from a dictionary — keys become column names
df_dict = pd.DataFrame({
    'Name':   ['Alice', 'Bob', 'Carol'],
    'Score':  [88, 95, 72],
    'Passed': [True, True, False]
})
df_dict
```

# 📌 Topic 12 — Sorting Values

`.sort_values(by='col')` sorts rows by the values in a column.

- Does **not** modify the original — returns a new DataFrame (catch it in a variable)
- `ascending=False` for Z→A or high→low
- Pass a **list** to `by` to sort by multiple columns
- Pass a **list** to `ascending` to set direction per column

```python
# Reload clean df for sorting demos
df = pd.read_csv('data/stocks.csv', index_col='Symbol')
df.head()
```

```python
# Sort by Name A → Z
df.sort_values(by='Name')
```

```python
# Sort by Price — highest first
df.sort_values(by='Price', ascending=False)
```

```python
# Sort by Sector A→Z, then by Price high→low within each sector
cols       = ['Sector', 'Price']
directions = [True, False]
df.sort_values(by=cols, ascending=directions)
```

# 📌 Topic 13 — Sorting Indexes

`.sort_index()` sorts the **row labels** (or column labels).

- `axis=0` or `axis='rows'` → sort the row index (default)
- `axis=1` or `axis='columns'` → sort the column index
- Like `sort_values`, it returns a new DataFrame — original unchanged

```python
# Sort row index A → Z (Symbol column becomes sorted)
df.sort_index()
```

```python
# Sort row index Z → A
df.sort_index(ascending=False)
```

```python
# Sort the COLUMN index A → Z
df.sort_index(axis=1)
```

```python
# Sort column index Z → A
df.sort_index(axis='columns', ascending=False)
```

```python
# Useful pattern: transpose then sort columns
df.transpose().sort_index(axis=1)
```

# 📌 Topic 14 — Resetting & Setting Indexes

- `reset_index()` → throws away the current index and replaces it with 0, 1, 2...
  The old index becomes a regular column.
- `set_index('col')` → promotes a column to be the new index

These two are often used together.

```python
# Current index is Symbol
print('Current index:', df.index.tolist()[:5])
```

```python
# reset_index() — Symbol moves into a regular column, index becomes 0,1,2...
df_reset = df.reset_index()
df_reset.head()
```

```python
# set_index() — promote Name column to be the new index
df_named = df_reset.set_index('Name')
df_named.head()
```

```python
# Now we can look up rows by company name
df_named.loc['Apple Inc']
```

# 📌 Topic 15 — Filtering Data

Filters let you select only the rows that match a condition — like WHERE in SQL or AutoFilter in Excel.

**Syntax:** `df[ condition ]`  where condition is True/False for each row.

- Save filters as variables so you can reuse and combine them
- `&` = AND (both conditions must be true)
- `|` = OR (at least one condition must be true)
- Chain `.loc[]` after a filter to also restrict by row label

```python
# Reload clean df
df = pd.read_csv('data/stocks.csv', index_col='Symbol')
```

```python
# A filter expression — produces True/False for each row
df['Price'] > 150
```

```python
# Apply the filter to the DataFrame — only matching rows returned
df[ df['Price'] > 150 ]
```

```python
# Save filters as named variables — much cleaner to read
HIGH_PRICE = 150
LOW_VOL    = 10_000_000

price_gt_150 = df['Price'] > HIGH_PRICE
vol_lt_10M   = df['Volume'] < LOW_VOL

# Apply a single filter
df[price_gt_150]
```

```python
# Filter: Volume < 10 million
df[vol_lt_10M]
```

```python
# Filter by comparing two columns — Price > Volume/1M (just for illustration)
price_gt_vol = df['Price'] > df['Volume'] / 1_000_000
df[price_gt_vol]
```

```python
# Count how many rows match a filter
df[price_gt_150].count()
```

```python
# AND filter — Price > 150 AND Volume < 10M
df[price_gt_150 & vol_lt_10M]
```

```python
# OR filter — Price > 150 OR Volume < 10M
df[price_gt_150 | vol_lt_10M]
```

```python
# Filter and then select specific columns
df[price_gt_150][['Name', 'Sector', 'Price']]
```

# 📌 Topic 16 — MultiIndex & Cross-Sections

A **MultiIndex** lets you have multiple levels of row labels — like a nested grouping.

Example: Rep → Portfolio → Sector

Steps to build one:
1. Load data
2. `zip()` the columns you want as levels into tuples
3. Build the index with `pd.MultiIndex.from_tuples()`
4. Assign it to `df.index`
5. Clean up — drop the columns that are now in the index
6. Name the levels with `.index.names`
7. Use `.xs()` to slice across any level

```python
# Step 1 — Load the portfolios data (normal flat index for now)
df_port = pd.read_csv('data/portfolios.csv')
df_port.head()
```

```python
# Step 2 — zip() creates a list of tuples from multiple columns
tuples = list(zip(df_port['Rep'], df_port['Portfolio'], df_port['Sector']))
print('First 3 tuples:')
print(tuples[:3])
```

```python
# Step 3 — Build the MultiIndex from the tuples
hier_index = pd.MultiIndex.from_tuples(tuples)
print(hier_index)
```

```python
# Step 4 — Assign the MultiIndex to the DataFrame
df_port.index = hier_index
df_port.head()
```

```python
# Step 5 — Drop columns that are now in the index
df_port.pop('Rep')
df_port.pop('Portfolio')
df_port.pop('Sector')
df_port.head()
```

```python
# Step 6 — Name the index levels
df_port.index.names = ['Rep', 'Portfolio', 'Sector']
df_port
```

```python
# Step 7 — Cross-section with .xs()
# All rows where Rep == John
df_port.xs(key='John', level='Rep')
```

```python
# All rows in Portfolio P5
df_port.xs(key='P5', level='Portfolio')
```

```python
# All rows in the Technology sector
df_port.xs(key='Technology', level='Sector')
```

```python
# Cross-section using multiple levels at once
keys   = ('Ringo', 'P4', 'Healthcare')
levels = ['Rep', 'Portfolio', 'Sector']
df_port.xs(key=keys, level=levels)
```
