## The Python `datetime` Library — Big Picture

Python's date and time functionality lives inside a library (also called a module) named **`datetime`**.

⚠️ **Common confusion**: The name `datetime` is used **twice**:
1. As the name of the **top-level library** (the container)
2. As the name of one of the **sub-libraries inside it** (the class)

Think of it like this:

```
datetime          ← top-level library (the toolbox)
├── datetime      ← sub-library: represents a full date + time
├── date          ← sub-library: represents just a date (no time)
├── time          ← sub-library: represents just a time (no date)
├── timedelta     ← sub-library: represents a duration / difference
├── timezone      ← sub-library: basic UTC offset timezone support
├── MAXYEAR       ← constant: the maximum year Python can represent (9999)
└── MINYEAR       ← constant: the minimum year Python can represent (1)
```

For more advanced timezone support (e.g., 'America/New_York'), we also use the **`pytz`** package (covered in Part 3).

## 1.1 Importing the Top-Level `datetime` Library

The simplest way to import is to bring in the entire top-level library at once.

You then access sub-libraries using **dot notation**: `datetime.datetime`, `datetime.MAXYEAR`, etc.

**Syntax:**
```python
import datetime

datetime.MAXYEAR    # → 9999
datetime.MINYEAR    # → 1
datetime.datetime   # → the datetime class inside the library
```

The two global **constants** tell us the range of years Python's datetime system can handle:
- `MINYEAR = 1` → January 1st, year 1 AD is the earliest representable date
- `MAXYEAR = 9999` → December 31st, year 9999 is the latest representable date

````python
# Suppress deprecation and future warnings so output stays clean
import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)
warnings.filterwarnings("ignore", category=FutureWarning)

# Import the top-level datetime library (the entire toolbox)
import datetime

# MAXYEAR: the latest year Python's datetime can represent
print("Maximum year:", datetime.MAXYEAR)   # → 9999

# MINYEAR: the earliest year Python's datetime can represent
print("Minimum year:", datetime.MINYEAR)   # → 1
````

---

## 1.2 Importing Sub-Libraries Directly

Instead of typing `datetime.datetime(...)` every time, you can import the specific class you need directly into your namespace.

**Syntax:**
```python
from <top-level-library> import <sub-library>
```

After this, you can use `datetime(...)` directly without the `datetime.` prefix.

**The three most commonly used sub-libraries:**

| Sub-library | What it represents | Example |
|---|---|---|
| `datetime` | A full date AND time | `datetime(2024, 6, 15, 9, 30, 0)` → June 15 2024, 09:30:00 |
| `time` | A time of day (no date) | `time(14, 30, 0)` → 14:30:00 |
| `timedelta` | A duration or difference | `timedelta(days=7)` → 7 days |

**Constructor signatures:**
```python
datetime(year, month, day, hour=0, minute=0, second=0, microsecond=0)
time(hour=0, minute=0, second=0, microsecond=0)
timedelta(days=0, seconds=0, microseconds=0, milliseconds=0, minutes=0, hours=0, weeks=0)
```

````python
# Import the datetime CLASS from the datetime LIBRARY
# After this, we can write datetime(...) instead of datetime.datetime(...)
from datetime import datetime

# Import the time class — represents a time of day without any date
from datetime import time

# Import timedelta — represents a duration or difference between two points in time
from datetime import timedelta

# --- Demonstrating each class ---

# Create a full datetime object: August 20, 2017 at midnight (00:00:00)
dt = datetime(2017, 8, 20)
print("datetime object:", dt)       # → 2017-08-20 00:00:00

# Create a time object: 14:30:00 (2:30 PM, no date component)
t = time(14, 30, 0)
print("time object:", t)            # → 14:30:00

# Create a timedelta of exactly 10 days
td = timedelta(days=10)
print("timedelta object:", td)      # → 10 days, 0:00:00
````

## 1.3 Getting the Current Date and Time

`datetime.now()` returns the **current local date and time** as a `datetime` object.

This is one of the most frequently used functions — useful for logging, timestamping events, and measuring elapsed time.

**Key methods for getting 'now':**

| Method | Returns | Example output |
|---|---|---|
| `datetime.now()` | Current local date and time | `2024-06-15 14:23:47.183421` |
| `datetime.utcnow()` | Current UTC date and time | `2024-06-15 08:53:47.183421` |
| `datetime.today()` | Same as `now()` (no timezone support) | `2024-06-15 14:23:47.183421` |

````python
# datetime.now() returns the current local date and time
now = datetime.now()
print("Current date and time:", now)

# You can access individual components of a datetime object using attributes
print("Year:       ", now.year)         # 4-digit year, e.g. 2024
print("Month:      ", now.month)        # 1–12
print("Day:        ", now.day)          # 1–31
print("Hour:       ", now.hour)         # 0–23
print("Minute:     ", now.minute)       # 0–59
print("Second:     ", now.second)       # 0–59
print("Microsecond:", now.microsecond)  # 0–999999

# .date() strips the time component and returns just the date
print("Date only:  ", now.date())       # e.g. 2024-06-15

# .time() strips the date component and returns just the time
print("Time only:  ", now.time())       # e.g. 14:23:47.183421
````

## 1.4 Date Arithmetic — Subtracting Two Datetimes

One of the most powerful features of the `datetime` library is that you can perform **arithmetic** on datetime objects — just like you would with numbers.

**Subtracting two datetime objects** always returns a **`timedelta`** object, which represents the difference.

```
datetime - datetime  →  timedelta
datetime + timedelta →  datetime
datetime - timedelta →  datetime
```

**The `timedelta` object has three key attributes:**

| Attribute | What it stores | Note |
|---|---|---|
| `.days` | Total number of whole days | Can be negative |
| `.seconds` | Remaining seconds (after whole days) | 0–86399 |
| `.microseconds` | Remaining microseconds | 0–999999 |

⚠️ **Important**: `.seconds` is NOT the total seconds — it's only the seconds remaining after the days are accounted for. To get total seconds, use `.total_seconds()`.

````python
# Get the current moment in time
now = datetime.now()

# Create a specific past date: August 20, 2017 at midnight
then = datetime(2017, 8, 20)

# Subtract the past datetime from now — this returns a timedelta object
# The result represents 'how long ago was August 20, 2017'
delta = now - then

# Print the full timedelta (shows as 'X days, H:MM:SS.ffffff')
print("Full delta:       ", delta)

# .days: the number of whole days in the difference
print("Days:             ", delta.days)

# .seconds: the remaining seconds AFTER whole days are accounted for
# This is NOT the total seconds — just the sub-day remainder
print("Remaining seconds:", delta.seconds)

# .total_seconds(): the TOTAL duration expressed entirely in seconds
# This is what you want when comparing durations numerically
print("Total seconds:    ", delta.total_seconds())

# Practical example: how many weeks is that?
print("Approximate weeks:", round(delta.days / 7, 1))
````

## 1.5 Timedeltas — Working with Durations

A `timedelta` is not just the result of subtraction — you can also **create one directly** and use it to shift dates forward or backward.

This is extremely useful for:
- Adding 30 days to today to find a deadline
- Subtracting 90 days to find the start of a quarter
- Checking if two events are within 7 days of each other

**Constructor:**
```python
timedelta(days=0, seconds=0, microseconds=0, milliseconds=0, minutes=0, hours=0, weeks=0)
```

All arguments are optional and default to 0. You can combine them freely.

**Timedelta arithmetic rules:**

| Operation | Example | Result type |
|---|---|---|
| `datetime + timedelta` | Add 12 days to a date | `datetime` |
| `datetime - timedelta` | Subtract 12 days from a date | `datetime` |
| `timedelta + timedelta` | Sum two durations | `timedelta` |
| `timedelta * int` | Multiply a duration | `timedelta` |
| `datetime - datetime` | Difference between two dates | `timedelta` |

````python
# Re-import timedelta for clarity (already imported above, but shown for reference)
from datetime import timedelta

# Create a starting date: January 7, 2013 at midnight
start = datetime(2013, 1, 7)
print("Start date:              ", start)

# --- Example 1: Add 12 days ---
# timedelta(12) is shorthand for timedelta(days=12)
twelve_days = timedelta(12)
result_1 = start + twelve_days
print("Start + 12 days:         ", result_1)    # → 2013-01-19 00:00:00

# --- Example 2: Add 2 days and 74447 seconds ---
# timedelta(2, 74447) means: 2 days + 74447 seconds
# 74447 seconds = 20 hours, 40 minutes, 47 seconds
result_2 = start + timedelta(2, 74447)
print("Start + 2 days 74447s:   ", result_2)    # → 2013-01-09 20:40:47

# --- Example 3: Named arguments are clearer ---
# Using keyword arguments makes the intent explicit
result_3 = start + timedelta(days=2, hours=20, minutes=40, seconds=47)
print("Same with named args:    ", result_3)    # → 2013-01-09 20:40:47

# --- Example 4: Subtracting a timedelta moves backward in time ---
result_4 = start - timedelta(weeks=2)
print("Start minus 2 weeks:     ", result_4)    # → 2012-12-24 00:00:00

# --- Example 5: Combining timedeltas ---
week  = timedelta(weeks=1)
extra = timedelta(days=3)
combined = week + extra
print("1 week + 3 days:         ", combined)    # → 10 days, 0:00:00

# --- Example 6: Negative timedeltas ---
# Subtracting a future date from an earlier date gives a negative timedelta
past   = datetime(2020, 1, 1)
future = datetime(2025, 1, 1)
negative_delta = past - future
print("Negative timedelta:      ", negative_delta)   # → -1827 days, 0:00:00
print("Is negative? days < 0:   ", negative_delta.days < 0)  # → True
````

## 1.6 Practical Example — Date Arithmetic in Banking

Let's apply what we've learned to a realistic scenario:

> A loan was issued on **March 1, 2023** with a **90-day repayment window**.  
> An early repayment discount applies if paid **30 days early**.  
> Calculate the due date, the discount deadline, and how many days remain from today.

````python
# Define the loan issue date
loan_issue_date = datetime(2023, 3, 1)

# Calculate the repayment due date: 90 days from issue
due_date = loan_issue_date + timedelta(days=90)
print("Loan issued on:    ", loan_issue_date.strftime('%B %d, %Y'))
print("Due date (90 days):", due_date.strftime('%B %d, %Y'))      # → May 30, 2023

# Early repayment deadline: 30 days before the due date
early_repay_deadline = due_date - timedelta(days=30)
print("Early repay by:    ", early_repay_deadline.strftime('%B %d, %Y'))  # → April 30, 2023

# Calculate how many days have passed since the loan was issued
today = datetime.now()
days_elapsed = (today - loan_issue_date).days
print(f"Days since issue:   {days_elapsed} days")

# Calculate days until the due date (can be negative if already past)
days_remaining = (due_date - today).days

# Use conditional logic to handle past vs future due dates
if days_remaining >= 0:
    print(f"Days until due:     {days_remaining} days remaining")
else:
    print(f"Loan overdue by:    {abs(days_remaining)} days")
````

# Part 2: Converting Datetime to String

---

## 2.1 Why Convert Datetimes to Strings?

A `datetime` object is great for arithmetic, but when you need to **display** a date to a user, write it to a file, or embed it in a report, you need a **formatted string**.

Python provides two complementary methods:

| Method | Direction | Example |
|---|---|---|
| `strftime` | datetime → string | `"Friday 21 September 2018"` |
| `strptime` | string → datetime | `"2018-09-21"` → `datetime(2018, 9, 21)` |

**Memory trick:**  
- `strftime` = **f**ormat (datetime **to** string)
- `strptime` = **p**arse (string **to** datetime)

---

## 2.2 Format Codes Reference

Format codes are **special placeholders** that get replaced with parts of the date/time. They always start with `%`.

Full reference: http://strftime.org/

### Date Codes

| Code | Meaning | Example |
|---|---|---|
| `%Y` | 4-digit year | `2018` |
| `%y` | 2-digit year | `18` |
| `%m` | Month as zero-padded number | `09` |
| `%B` | Full month name | `September` |
| `%b` | Abbreviated month name | `Sep` |
| `%d` | Day as zero-padded number | `21` |
| `%A` | Full weekday name | `Friday` |
| `%a` | Abbreviated weekday name | `Fri` |
| `%w` | Weekday as number (0=Sunday) | `5` |
| `%j` | Day of year (001–366) | `264` |

### Time Codes

| Code | Meaning | Example |
|---|---|---|
| `%H` | Hour (24-hour clock, zero-padded) | `14` |
| `%I` | Hour (12-hour clock, zero-padded) | `02` |
| `%M` | Minutes (zero-padded) | `30` |
| `%S` | Seconds (zero-padded) | `05` |
| `%p` | AM or PM | `PM` |

### Compound Codes (shortcuts)

| Code | Equivalent | Example |
|---|---|---|
| `%c` | Full locale datetime | `Fri Sep 21 14:30:05 2018` |
| `%x` | Locale's date | `09/21/18` |
| `%X` | Locale's time | `14:30:05` |

````python
# Create a sample datetime: September 21, 2018 at midnight
stamp = datetime(2018, 9, 21)
print("Raw datetime object:", stamp)
print()

# --- Method 1: str() converts to a basic ISO 8601 string ---
# ISO 8601 format: YYYY-MM-DD HH:MM:SS
# This is the default string representation — good for storage and logging
iso_string = str(stamp)
print("str() output:          ", iso_string)          # → '2018-09-21 00:00:00'

# --- Method 2: strftime with ISO date format ---
# %Y = 4-digit year, %m = 2-digit month, %d = 2-digit day
iso_date = stamp.strftime('%Y-%m-%d')
print("ISO date format:       ", iso_date)             # → '2018-09-21'

# --- Method 3: Human-readable long format ---
# %A = full weekday, %d = day number, %B = full month, %Y = 4-digit year
long_format = stamp.strftime('%A %d %B %Y')
print("Long human format:     ", long_format)          # → 'Friday 21 September 2018'

# --- More format examples ---

# Short date (common in US): MM/DD/YYYY
us_format = stamp.strftime('%m/%d/%Y')
print("US date format:        ", us_format)            # → '09/21/2018'

# Short date (common in UK/India): DD/MM/YYYY
uk_format = stamp.strftime('%d/%m/%Y')
print("UK/India date format:  ", uk_format)            # → '21/09/2018'

# Just the month and year (useful for report headers)
month_year = stamp.strftime('%B %Y')
print("Month and year only:   ", month_year)           # → 'September 2018'

# Abbreviated format
abbreviated = stamp.strftime('%a, %b %d \'%y')
print("Abbreviated format:    ", abbreviated)          # → 'Fri, Sep 21 '18'

# Day of the year (useful for financial calendars)
day_of_year = stamp.strftime('Day %j of %Y')
print("Day of year:           ", day_of_year)          # → 'Day 264 of 2018'
````

---

## 2.3 Formatting Datetimes with Time Components

When your datetime object includes a time (not just midnight), you can format that too.

````python
# Create a datetime with both date AND time components
# Parameters: year, month, day, hour, minute, second
meeting = datetime(2024, 3, 15, 14, 30, 45)
print("Raw datetime:", meeting)
print()

# 24-hour time format with seconds
# %H = hour (00-23), %M = minute (00-59), %S = second (00-59)
time_24h = meeting.strftime('%H:%M:%S')
print("24-hour time:          ", time_24h)     # → '14:30:45'

# 12-hour time format with AM/PM
# %I = hour (01-12), %p = AM or PM
time_12h = meeting.strftime('%I:%M %p')
print("12-hour time:          ", time_12h)     # → '02:30 PM'

# Full datetime: date + time together
full_format = meeting.strftime('%A, %B %d %Y at %I:%M %p')
print("Full meeting time:     ", full_format)  # → 'Friday, March 15 2024 at 02:30 PM'

# Compact sortable format (useful for filenames and logs)
# Dates formatted this way sort correctly in alphabetical order
sortable = meeting.strftime('%Y%m%d_%H%M%S')
print("Sortable/filename fmt: ", sortable)     # → '20240315_143045'
````

## 2.4 Parsing Strings Back to Datetimes — `strptime`

`strptime` does the **reverse** of `strftime`: it takes a string and a format pattern, and converts the string back into a `datetime` object.

This is essential when reading dates from:
- CSV files
- Databases
- User input
- API responses

**Syntax:** `datetime.strptime(date_string, format_string)`

````python
# --- strptime: parse a string into a datetime object ---

# Example 1: Parse a standard date string
# The format pattern must exactly match the structure of the string
date_str_1 = "2024-06-15"
parsed_1 = datetime.strptime(date_str_1, '%Y-%m-%d')
print("Parsed ISO date:       ", parsed_1)      # → 2024-06-15 00:00:00
print("Type:                  ", type(parsed_1)) # → <class 'datetime.datetime'>

# Example 2: Parse a human-readable date
date_str_2 = "21 September 2018"
parsed_2 = datetime.strptime(date_str_2, '%d %B %Y')
print("Parsed long date:      ", parsed_2)      # → 2018-09-21 00:00:00

# Example 3: Parse date WITH time
date_str_3 = "15/03/2024 14:30:00"
parsed_3 = datetime.strptime(date_str_3, '%d/%m/%Y %H:%M:%S')
print("Parsed date+time:      ", parsed_3)      # → 2024-03-15 14:30:00

# Example 4: Once parsed, you can do arithmetic on it
# Add 30 days to the parsed date
thirty_days_later = parsed_3 + timedelta(days=30)
print("30 days later:         ", thirty_days_later.strftime('%d %B %Y'))  # → 14 April 2024
````

# Part 3: Timezones

---

## 3.1 Why Timezones Are Important

By default, datetime objects in Python are **timezone-naive** — they don't know which timezone they represent. This causes problems in global applications:

- A transaction timestamped at `14:30` — is that London time? New York? Mumbai?
- If you compare two timestamps from different timezones, the comparison will be wrong
- Financial systems must be precise about exactly when something happened globally

**Naive vs. Aware datetimes:**

| Type | Description | Example |
|---|---|---|
| **Naive** | No timezone info | `datetime(2024, 6, 15, 14, 30)` |
| **Aware** | Has timezone info | `datetime(2024, 6, 15, 14, 30, tzinfo=...)` |

---

## 3.2 The `pytz` Package

Python's built-in timezone support is limited. The **`pytz`** package (Python Timezone) provides:
- Access to all IANA timezone names (e.g., `'America/New_York'`, `'Asia/Kolkata'`)
- Correct handling of Daylight Saving Time (DST)
- Easy conversion between timezones

**Installation** (if not already installed):
```bash
pip install pytz
```

**Common timezone names:**

| City | pytz timezone string |
|---|---|
| London | `'Europe/London'` |
| New York | `'America/New_York'` |
| Los Angeles | `'America/Los_Angeles'` |
| Mumbai / India | `'Asia/Kolkata'` |
| Singapore | `'Asia/Singapore'` |
| Tokyo | `'Asia/Tokyo'` |
| Sydney | `'Australia/Sydney'` |
| UTC | `'UTC'` |

````python
# Import the pytz library for timezone support
import pytz

# --- Step 1: Create a timezone object ---
# pytz.timezone() returns a timezone object for the named zone
# Timezone names follow the IANA database format: 'Region/City'
tz_new_york = pytz.timezone('America/New_York')
print("New York timezone object:", tz_new_york)

tz_london  = pytz.timezone('Europe/London')
print("London timezone object:  ", tz_london)

tz_india   = pytz.timezone('Asia/Kolkata')
print("India timezone object:   ", tz_india)

print()

# --- Step 2: Get the current time in a specific timezone ---
# Pass the timezone object to datetime.now() to get timezone-aware datetime
now_new_york = datetime.now(tz_new_york)
now_london   = datetime.now(tz_london)
now_india    = datetime.now(tz_india)

# Print each with timezone info included in the format string
# %Z = timezone abbreviation (e.g., EST, GMT, IST)
# %z = UTC offset (e.g., -0500, +0530)
print("New York time:", now_new_york.strftime('%Y-%m-%d %H:%M:%S %Z (%z)'))
print("London time:  ", now_london.strftime('%Y-%m-%d %H:%M:%S %Z (%z)'))
print("India time:   ", now_india.strftime('%Y-%m-%d %H:%M:%S %Z (%z)'))
````

---

## 3.3 Listing All Supported Timezones

`pytz.all_timezones` returns a **list** of all supported IANA timezone names.

There are **~600 timezone strings** in total. You can search through them to find the one you need.

**Format of timezone names:**
- `Continent/City` → `America/New_York`, `Asia/Kolkata`, `Europe/Paris`
- `UTC` → Universal Coordinated Time (no offset)
- Some historical or special names → `US/Eastern`, `GMT`, `Etc/GMT+5`

⚠️ Note: `'America/Salem'` does **not** exist as a timezone — Salem is in the `America/New_York` or `America/Los_Angeles` zone depending on location.

````python
# pytz.all_timezones returns a list of ALL supported timezone name strings
all_zones = pytz.all_timezones

# Display the total count
print(f"Total number of timezones: {len(all_zones)}")
print()

# Show the first 10 timezones (alphabetical order)
print("First 10 timezones:")
for zone in all_zones[:10]:
    print("  ", zone)

print("  ...")

# Show the last 5 timezones
print("Last 5 timezones:")
for zone in all_zones[-5:]:
    print("  ", zone)
````

````python
# --- Searching for a specific timezone ---
# Use a list comprehension to filter timezones containing a keyword

# Search all timezones that contain 'America'
# The 'in' operator checks if the keyword is a substring of the zone name
america_zones = [zone for zone in pytz.all_timezones if 'America' in zone]
print(f"'America' timezones ({len(america_zones)} found):")
for zone in america_zones[:15]:  # Print first 15 for brevity
    print("  ", zone)
print("  ...")
print()

# Search for 'India' related timezones
india_zones = [zone for zone in pytz.all_timezones if 'India' in zone or 'Kolkata' in zone]
print(f"India timezones ({len(india_zones)} found):")
for zone in india_zones:
    print("  ", zone)
print()

# Search for 'Salem' — note: it does NOT exist as a standalone timezone!
salem_zones = [zone for zone in pytz.all_timezones if 'Salem' in zone]
if salem_zones:
    print("Salem timezones found:", salem_zones)
else:
    # Salem, Oregon is in the US/Pacific zone
    # Salem, Massachusetts is in the US/Eastern zone
    print("'America/Salem' does NOT exist in pytz.")
    print("Salem, Oregon → use 'America/Los_Angeles'")
    print("Salem, Massachusetts → use 'America/New_York'")
````

## 3.4 Converting Between Timezones

Once you have a timezone-aware datetime, you can convert it to any other timezone using `.astimezone()`.

This is critical for global financial systems where a transaction time in New York needs to be recorded in UTC and displayed in the user's local timezone.

````python
# --- Converting between timezones ---

# Step 1: Create a timezone-aware datetime in UTC
# UTC is the reference point — all other timezones are offsets from UTC
utc_zone  = pytz.timezone('UTC')
utc_time  = datetime.now(utc_zone)
print("UTC time:            ", utc_time.strftime('%Y-%m-%d %H:%M:%S %Z'))

# Step 2: Convert UTC to New York time (UTC-5 in winter, UTC-4 in summer/DST)
# .astimezone() converts the datetime to the target timezone
ny_zone   = pytz.timezone('America/New_York')
ny_time   = utc_time.astimezone(ny_zone)
print("New York time:       ", ny_time.strftime('%Y-%m-%d %H:%M:%S %Z'))

# Step 3: Convert to India time (UTC+5:30)
india_zone  = pytz.timezone('Asia/Kolkata')
india_time  = utc_time.astimezone(india_zone)
print("India (IST) time:    ", india_time.strftime('%Y-%m-%d %H:%M:%S %Z'))

# Step 4: Convert to Tokyo time (UTC+9)
tokyo_zone  = pytz.timezone('Asia/Tokyo')
tokyo_time  = utc_time.astimezone(tokyo_zone)
print("Tokyo time:          ", tokyo_time.strftime('%Y-%m-%d %H:%M:%S %Z'))

# Step 5: Convert to Sydney time (UTC+10 or UTC+11 with DST)
sydney_zone = pytz.timezone('Australia/Sydney')
sydney_time = utc_time.astimezone(sydney_zone)
print("Sydney time:         ", sydney_time.strftime('%Y-%m-%d %H:%M:%S %Z'))

print()
print("All of the above represent the SAME moment in time, just displayed in different zones.")
````

## 3.5 Localizing a Naive Datetime

If you have an existing naive datetime and want to make it timezone-aware, use `pytz.localize()` — **not** `replace(tzinfo=...)`.

⚠️ **Never use `replace(tzinfo=...)` with pytz** — it doesn't handle DST correctly.  
✅ **Always use `tz.localize(naive_datetime)`** to attach a timezone to a naive datetime.

````python
# --- Localizing a naive datetime ---

# This is a naive datetime — it has no timezone information
naive_dt = datetime(2024, 6, 15, 14, 30, 0)
print("Naive datetime:         ", naive_dt)
print("Timezone info (None):   ", naive_dt.tzinfo)   # → None (no timezone)
print()

# Use tz.localize() to attach a timezone to the naive datetime
# This says: 'This datetime represents a time IN New York'
eastern_zone = pytz.timezone('America/New_York')
aware_dt = eastern_zone.localize(naive_dt)
print("Aware datetime (NY):    ", aware_dt)
print("Timezone info:          ", aware_dt.tzinfo)   # → America/New_York
print()

# Now convert to UTC to store in a database (always store in UTC)
utc_zone = pytz.timezone('UTC')
stored_utc = aware_dt.astimezone(utc_zone)
print("Converted to UTC:       ", stored_utc)

# Convert to India time for display to an Indian user
india_zone = pytz.timezone('Asia/Kolkata')
display_india = aware_dt.astimezone(india_zone)
print("Display in India (IST): ", display_india.strftime('%d %b %Y, %I:%M %p %Z'))
````

## Key Rules to Remember

1. **`datetime - datetime = timedelta`** (always)
2. **`datetime + timedelta = datetime`** (always)
3. **`.seconds` ≠ total seconds** — use `.total_seconds()` for full duration in seconds
4. **Never use `replace(tzinfo=...)` with pytz** — always use `tz.localize()`
5. **Store dates in UTC, display in local timezone** — the golden rule for any multi-timezone system
6. **`strftime` formats TO string, `strptime` parses FROM string** (f = format, p = parse)
