# 📊 Introduction to Plotly

---

## What is Plotly?

**Plotly** is a Python graphing library that produces **interactive, publication-quality charts**. Unlike matplotlib (which creates static images), Plotly charts are:

- 🖱️ **Interactive** — hover to see values, click to toggle traces, zoom and pan
- 🌐 **Web-ready** — renders as HTML/JavaScript in Jupyter and browsers
- 🎨 **Beautiful by default** — polished design without manual styling
- 📱 **Responsive** — works on desktop and mobile

---

## Two APIs — Which to Use?

Plotly provides two ways to create charts:

| API | Module | Best For |
|---|---|---|
| **Plotly Express** (`px`) | `plotly.express` | Quick, high-level charts with minimal code |
| **Graph Objects** (`go`) | `plotly.graph_objs` | Full control over every chart element |

**Rule of thumb:** Start with Plotly Express. Switch to Graph Objects when you need fine-grained control.

# Part 1: Your First Plotly Chart

---

## 1.1 The Three-Step Pattern

Every Plotly chart follows the same structure:

```python
# Step 1: Create a Figure (the container)
fig = go.Figure()

# Step 2: Add Traces (the data series to plot)
fig.add_trace(go.Scatter(y=[1, 2, 3]))   # a line series
fig.add_trace(go.Bar(y=[3, 1, 2]))       # a bar series

# Step 3: Update Layout (title, axes, size, etc.)
fig.update_layout(title='My Chart')

# Step 4: Show it
fig.show()
```

**Key concepts:**
- **Figure**: the outer container — holds all traces and layout settings
- **Trace**: one data series — a line, bar series, scatter plot, etc.
- **Layout**: non-data visual properties — title, axis labels, size, colours

## 1.2 `go.Figure` vs `go.FigureWidget`

| Class | Output | Use When |
|---|---|---|
| `go.Figure` | Static Plotly chart (HTML) | Standard charts, export to HTML |
| `go.FigureWidget` | Interactive Jupyter widget | When you want Python callbacks on chart events |

````python
# ──────────────────────────────────────────────────────
# Using go.Figure — the standard chart container
# ──────────────────────────────────────────────────────

# Create an empty Figure object — the container for traces and layout
fig = go.Figure()

# Add a Scatter trace (displayed as a line by default)
# y=[2, 1, 4, 3]: the y-axis values; x-axis defaults to [0, 1, 2, 3]
fig.add_trace(go.Scatter(
    y=[2, 1, 4, 3],          # y data points
    name='Line Series',       # name shown in the legend
    mode='lines+markers'      # draw both lines and marker dots
))

# Add a Bar trace on the same figure
# Both traces appear on the same chart — Plotly overlays them
fig.add_trace(go.Bar(
    y=[1, 4, 3, 2],           # y data points
    name='Bar Series'          # name shown in the legend
))

# update_layout: set chart-level properties (not data-related)
# title: the text that appears at the top of the chart
fig.update_layout(
    title='Hello Plotly — Combined Line and Bar Chart',
    xaxis_title='Index',       # label for the x-axis
    yaxis_title='Value'        # label for the y-axis
)

# Render the chart in the notebook
# In Jupyter: displays inline HTML chart
# Standalone: opens in the browser
fig.show()
````

````python
# ──────────────────────────────────────────────────────
# Using go.FigureWidget — for interactive Jupyter widgets
# ──────────────────────────────────────────────────────
# FigureWidget is a Jupyter widget — it can respond to Python callbacks
# NOTE: FigureWidget requires 'anywidget' package installed in the Jupyter environment
#       If running as a standalone script, it falls back gracefully
# Otherwise identical to go.Figure in terms of API
try:
    fig_w = go.FigureWidget()   # Try creating FigureWidget (needs anywidget)

    # Add the same traces as before
    fig_w.add_trace(go.Scatter(
        y=[2, 1, 4, 3],
        name='Line',
        mode='lines+markers'
    ))

    fig_w.add_trace(go.Bar(
        y=[1, 4, 3, 2],
        name='Bar'
    ))

    # update_layout works identically on FigureWidget
    fig_w.update_layout(title='Hello FigureWidget')
    fig_w.show()

except ImportError:
    # FigureWidget requires the 'anywidget' package in Jupyter
    # Fall back to go.Figure — identical output, just without widget interactivity
    print("FigureWidget requires 'anywidget' package. Using go.Figure instead.")
    print("Install with: pip install anywidget")
    fig_w = go.Figure()
    fig_w.add_trace(go.Scatter(y=[2, 1, 4, 3], name='Line', mode='lines+markers'))
    fig_w.add_trace(go.Bar(y=[1, 4, 3, 2], name='Bar'))
    fig_w.update_layout(title='Hello FigureWidget (fallback to Figure)')
    fig_w.show()
````

# Part 2: Tables in Plotly

## 2.1 Why Plotly Tables?

Plotly tables are **interactive** — they're scrollable, hoverable, and can be exported to HTML alongside your charts. They're ideal for embedding alongside Plotly charts in dashboards.

**Key structure:**
```python
go.Table(
    header = dict(values=['Col1', 'Col2']),   # column headers
    cells  = dict(values=[col1_data, col2_data])  # data per column
)
```

**Important:** `cells.values` is a **list of columns** (not rows):
```python
# CORRECT — list of column arrays:
cells = dict(values=[[row0_col0, row1_col0], [row0_col1, row1_col1]])

# This is different from pandas where you think in rows!
```

## 2.2 Build the Dataset

````python
# Build an S&P 500 sector summary dataset in memory
# In real usage: df_raw = pd.read_csv('../Data/constituents.csv', index_col='Symbol')
from io import StringIO

sp500_data = """Sector,Price,52 Week High,52 Week Low,CIK
Communication Services,42.15,51.30,33.20,1018724
Consumer Discretionary,85.32,102.50,68.40,899689
Consumer Staples,63.44,71.20,55.80,1170010
Energy,54.21,68.90,39.50,895126
Financials,72.18,88.40,58.30,1326160
Health Care,91.75,110.20,75.60,1090012
Industrials,68.93,82.10,56.70,731642
Information Technology,156.42,190.80,122.30,789019
Materials,47.88,58.20,38.90,66740
Real Estate,38.65,46.70,30.10,1040971
Utilities,52.34,61.80,43.50,1081316"""

# Read the CSV-formatted string into a DataFrame
# index_col='Sector': use the Sector column as the row index
df = pd.read_csv(StringIO(sp500_data), index_col='Sector')

print("S&P 500 Sector Summary Dataset:")
print(df)
print(f"\nShape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
````

````python
# ──────────────────────────────────────────────────────
# Basic Table — minimal configuration
# ──────────────────────────────────────────────────────

# Build the list of column names for the header row
col_names = list(df.columns)    # ['Price', '52 Week High', '52 Week Low', 'CIK']

# header_data: a dict defining the header row appearance
# 'values': the text to display in each header cell
header_data = dict(values=col_names)

# cell data must be a LIST OF COLUMNS (not rows!)
# Each element in the list is one column's data
# df['Price'] is a Series of all prices, etc.
data = [df['CIK'], df['Price'], df['52 Week High'], df['52 Week Low']]
cell_data = dict(values=data)

# go.Table creates a table trace
# header: configures the top header row
# cells:  configures all the data rows below
tbl = go.Table(header=header_data, cells=cell_data)

# Wrap the table in a Figure and display it
fig = go.Figure(data=tbl)
fig.update_layout(title='S&P 500 Sectors — Basic Table')
fig.show()
````

````python
# ──────────────────────────────────────────────────────
# Styled Table — add colours, alignment, fonts
# ──────────────────────────────────────────────────────

# Column names for the header (same as before)
col_names = list(df.columns)

# Header with styling properties
header_data = dict(
    values=col_names,           # text for each header cell
    fill_color='paleturquoise', # background colour of the header row
    align='left'                # text alignment: 'left', 'center', 'right'
)

# Cell data with styling
data = [df['CIK'], df['Price'], df['52 Week High'], df['52 Week Low']]
cell_data = dict(
    values=data,           # the data arrays (one per column)
    fill_color='lavender', # background colour of all data cells
    align='left'           # text alignment in cells
)

# Create the styled table
tbl = go.Table(header=header_data, cells=cell_data)

fig = go.Figure(data=tbl)

# update_layout: set overall chart dimensions in pixels
fig.update_layout(
    title='S&P 500 Sectors — Styled Table',
    width=800,   # chart width in pixels
    height=400   # chart height in pixels
)
fig.show()
````

````python
# ──────────────────────────────────────────────────────
# Fully Formatted Table — include the index, custom fonts, number formatting
# ──────────────────────────────────────────────────────

# Start with column names and prepend 'Sector' for the index column
col_names = list(df.columns)
col_names.insert(0, 'Sector')  # insert 'Sector' at position 0 (front)

# Configure header with full styling
header_data = dict(
    values=col_names,                   # column names including 'Sector'
    fill_color='paleturquoise',          # header background colour
    align=['left', 'center'],            # left for 1st col, center for rest
    font=dict(color='darkslategray', size=13),  # header text style
    height=40                            # row height in pixels
)

# Prepare data: include the index (Sector names) as the first column
data = [df['CIK'], df['Price'], df['52 Week High'], df['52 Week Low']]
data.insert(0, list(df.index))   # insert sector names as the first data column

# Configure cells with full styling
cell_data = dict(
    values=data,                         # data arrays including sector names
    fill_color='lavender',               # cell background colour
    align=['left', 'center'],            # left for sector, center for numbers
    font_size=12,                        # font size for all cell text
    format=[None, ',.2f'],               # None = no format, ',.2f' = 2 decimal places with comma
    height=30                            # row height in pixels
)

tbl = go.Table(header=header_data, cells=cell_data)

fig = go.Figure(data=tbl)
fig.update_layout(
    title='S&P 500 Sectors — Full Formatted Table',
    width=800,
    height=450
)
fig.show()
````

````python
# ──────────────────────────────────────────────────────
# Alternating Row Colours — professional look
# ──────────────────────────────────────────────────────

# Define the colours for different parts of the table
headerColor = 'grey'       # dark header
rowEvenColor = 'lightgrey' # even rows (index 0, 2, 4, ...)
rowOddColor  = 'white'     # odd rows  (index 1, 3, 5, ...)

# Build column names with index
col_names = list(df.columns)
col_names.insert(0, 'Sector')

# Header configuration with dark background and white text
header_vals = dict(
    values=col_names,
    line_color='darkslategray',              # border around header cells
    fill_color=headerColor,                  # grey header background
    align=['left', 'center'],
    font=dict(color='white', size=13),       # white text on dark background
    height=40
)

# Prepare data arrays
data = [df['CIK'], df['Price'], df['52 Week High'], df['52 Week Low']]
data.insert(0, list(df.index))

# Build alternating row colours dynamically
# df.index contains the Sector names; enumerate their positions with get_loc
# %2==0: even-indexed rows get 'lightgrey'; odd rows get 'white'
# The pattern is repeated for each column (len(df.columns) times) because
# Plotly expects one colour per CELL not per row
fill_colours = [
    [rowEvenColor if i % 2 == 0 else rowOddColor
     for i in range(len(df.index))] * len(df.columns)
]

# Cell configuration with alternating colours
cell_vals = dict(
    values=data,
    fill_color=fill_colours,   # alternating colour list per row
    align=['left', 'center'],
    font_size=12,
    format=[None, ',.2f'],
    height=30
)

tbl = go.Table(header=header_vals, cells=cell_vals)

fig = go.Figure(data=tbl)
fig.update_layout(
    title='S&P 500 Sectors — Alternating Row Colours',
    width=800,
    height=500
)
fig.show()
````

## 3.1 `go.Scatter()` — The Workhorse of Plotly

`go.Scatter()` is used for lines, scatter plots, and combinations of both. The `mode` parameter controls what is drawn:

| `mode` value | What is drawn |
|---|---|
| `'lines'` | Continuous line only |
| `'markers'` | Dots/markers only (scatter plot) |
| `'lines+markers'` | Both line and dots |
| `'text'` | Labels only |
| `'lines+markers+text'` | All three |

**Key parameters:**

| Parameter | Type | Description |
|---|---|---|
| `x` | list / Series | X-axis values |
| `y` | list / Series | Y-axis values |
| `mode` | str | What to draw (lines, markers, text) |
| `name` | str | Legend label |
| `marker` | dict | Dot styling (color, size, symbol) |
| `line` | dict | Line styling (color, width, dash) |
| `opacity` | float | Transparency (0.0 = invisible, 1.0 = solid) |
| `text` | list | Hover text per data point |

````python
# ──────────────────────────────────────────────────────
# Line plot: 52 Week High and Low by S&P 500 Sector
# ──────────────────────────────────────────────────────

# Trace 1: 52 Week High — a solid green line
# x: the sector names (index of df), shown on x-axis
# y: the 52 Week High value for each sector
# mode='lines': draw only a continuous line (no dots)
# name: appears in the legend
# marker=dict(color=...): RGBA colour with opacity
#   rgba(R, G, B, alpha) where R/G/B are 0-255 and alpha is 0.0-1.0
trace1 = go.Scatter(
    x=list(df.index),             # sector names as x-axis categories
    y=df['52 Week High'],          # 52-week high prices as y-axis values
    mode='lines',                  # draw a line (no markers)
    name='52 Wk High',             # legend label
    marker=dict(color='rgba(16, 112, 2, 0.8)')  # green, 80% opacity
)

# Trace 2: 52 Week Low — a purple line with markers
# mode='lines+markers': draw both a line AND dots at each data point
# This makes individual data points easy to see and hover over
trace2 = go.Scatter(
    x=list(df.index),             # same x-axis categories
    y=df['52 Week Low'],           # 52-week low prices as y-axis values
    mode='lines+markers',          # line AND dots
    name='52 Wk Low',              # legend label
    marker=dict(color='rgba(80, 26, 80, 0.8)')  # purple, 80% opacity
)

# Combine both traces into a list
data = [trace1, trace2]

# Layout: a dictionary of chart-level settings
layout = dict(
    title='52 Week High / Low by Sector — S&P 500',
    xaxis=dict(
        title='Sector',        # x-axis label
        ticklen=5,             # length of tick marks in pixels
        zeroline=False         # hide the zero line on x-axis
    ),
    yaxis=dict(
        title='Price (USD)',   # y-axis label
    )
)

# Create Figure by passing data and layout together
fig = go.Figure(data=data, layout=layout)
fig.show()
````

````python
# ──────────────────────────────────────────────────────
# Enhanced line plot: using update_traces and update_layout
# ──────────────────────────────────────────────────────

# Three traces: Price, 52 Wk High, 52 Wk Low
trace_price = go.Scatter(
    x=list(df.index),
    y=df['Price'],
    mode='lines+markers',
    name='Current Price',
    line=dict(color='royalblue', width=2),       # line style
    marker=dict(symbol='circle', size=8)          # marker style
)

trace_high = go.Scatter(
    x=list(df.index),
    y=df['52 Week High'],
    mode='lines',
    name='52 Wk High',
    line=dict(color='green', width=1, dash='dash')  # dashed green line
)

trace_low = go.Scatter(
    x=list(df.index),
    y=df['52 Week Low'],
    mode='lines',
    name='52 Wk Low',
    line=dict(color='red', width=1, dash='dot')     # dotted red line
)

# Create Figure with all three traces
fig = go.Figure(data=[trace_price, trace_high, trace_low])

# update_layout: comprehensive chart customisation
fig.update_layout(
    title=dict(text='S&P 500 Sector Prices vs 52-Week Range', font=dict(size=16)),
    xaxis=dict(title='Sector', tickangle=-30),  # tilt x labels to avoid overlap
    yaxis=dict(title='Price (USD)'),
    legend=dict(x=0.01, y=0.99),               # legend position (top-left)
    plot_bgcolor='white',                       # white chart background
    paper_bgcolor='white',                      # white outer background
    width=850,
    height=500
)

fig.show()
````

# Part 4: Bar Charts

---

## 4.1 `go.Bar()` — Grouped and Stacked Bars

`go.Bar()` creates bar charts. Multiple bar traces can be displayed as:
- **Grouped** (`barmode='group'`): bars side by side
- **Stacked** (`barmode='stack'`): bars on top of each other
- **Overlay** (`barmode='overlay'`): bars on top of each other without stacking

**Key parameters:**

| Parameter | Description |
|---|---|
| `x` | Categories on x-axis |
| `y` | Bar heights |
| `name` | Legend label |
| `marker.color` | Fill colour (RGBA or named) |
| `marker.line` | Border around each bar |
| `opacity` | Overall trace opacity |

````python
# ──────────────────────────────────────────────────────
# Grouped Bar Chart: 52 Week High vs Low by Sector
# ──────────────────────────────────────────────────────

# Trace 1: 52 Week High bars — pink/purple fill
trace1 = go.Bar(
    x=list(df.index),              # x categories: sector names
    y=df['52 Week High'],           # bar heights: 52-week high prices
    name='52 Week High',            # legend label
    marker=dict(
        color='rgba(255, 174, 255, 0.5)',    # semi-transparent pink fill
        line=dict(
            color='rgb(0, 0, 0)',   # black border around each bar
            width=1.5               # border line width in pixels
        )
    )
)

# Trace 2: 52 Week Low bars — yellow fill
trace2 = go.Bar(
    x=list(df.index),              # same x categories
    y=df['52 Week Low'],            # bar heights: 52-week low prices
    name='52 Week Low',             # legend label
    marker=dict(
        color='rgba(255, 255, 128, 0.5)',    # semi-transparent yellow fill
        line=dict(
            color='rgb(0, 0, 0)',   # black border
            width=1.5
        )
    )
)

# Combine traces in a list
data = [trace1, trace2]

# go.Layout: defines the layout object (alternative to update_layout)
# barmode='group': place bars SIDE BY SIDE for each x category
# Other options: 'stack' (stacked), 'overlay' (overlapping), 'relative'
layout = go.Layout(
    barmode='group',
    title='Mean 52 Wk High & Low by Sector — S&P 500',
    xaxis=dict(title='Sector', tickangle=-30),
    yaxis=dict(title='Price (USD)'),
    width=850,
    height=450
)

# Create Figure with both data and layout
fig = go.Figure(data=data, layout=layout)
fig.show()
````

````python
# ──────────────────────────────────────────────────────
# Stacked Bar Chart — alternative to grouped
# ──────────────────────────────────────────────────────

# Same traces, different layout barmode
fig_stacked = go.Figure(data=[trace1, trace2])

# barmode='stack': bars stacked on top of each other
# Total height = sum of all traces for that category
fig_stacked.update_layout(
    barmode='stack',
    title='52 Wk High + Low Stacked by Sector (Stacked Mode)',
    xaxis=dict(title='Sector', tickangle=-30),
    yaxis=dict(title='Combined Price (USD)'),
    width=850,
    height=450
)
fig_stacked.show()
````

# Part 5: Time Series

---

## 5.1 Why Plotly for Time Series?

Plotly is particularly powerful for time series because:
- **Range slider**: lets users zoom into a time period interactively
- **Range selector buttons**: quick jumps to 1M, 3M, 6M, 1Y, All time
- **Hover tooltips**: show exact date and value on cursor hover
- **Datetime-aware x-axis**: correctly spaces irregular time intervals

````python
# Build an AAPL-like stock price dataset in memory
# In real usage: df = pd.read_csv('../Data/AAPL.csv', index_col='Date', parse_dates=True)
import pandas as pd
import numpy as np
from io import StringIO

# Generate synthetic daily stock prices using a random walk
# This simulates a realistic stock price series
np.random.seed(42)                          # seed for reproducibility
dates = pd.date_range('2015-01-01', '2017-12-31', freq='B')  # business days only
n = len(dates)

# Simulate price series: start at 100, add random daily returns
price = 100 + np.cumsum(np.random.randn(n) * 1.5)  # cumulative random walk
high  = price + np.abs(np.random.randn(n) * 2)      # high is always above price
low   = price - np.abs(np.random.randn(n) * 2)      # low is always below price
close = price + np.random.randn(n) * 0.5             # close near price

# Build the DataFrame from the simulated data
df_stock = pd.DataFrame({
    'High': high,
    'Low':  low,
    'Close': close
}, index=dates)
df_stock.index.name = 'Date'

print("Stock price dataset:")
print(df_stock.head(10).round(2))
print(f"\nDate range: {df_stock.index[0].date()} to {df_stock.index[-1].date()}")
print(f"Total trading days: {len(df_stock)}")
````

````python
# ──────────────────────────────────────────────────────
# Method 1: Plotly Express — quickest approach
# ──────────────────────────────────────────────────────
# px.line: creates a line chart in one function call
# x=df_stock.index: the datetime index becomes the x-axis
# y='High': the column to plot on the y-axis
print("=== Plotly Express (one-liner) ===")
fig_px = px.line(
    df_stock,                 # the DataFrame
    x=df_stock.index,         # x-axis: the DatetimeIndex
    y='High',                 # y-axis: the 'High' column
    title='Stock High Price (2015–2017) — Plotly Express',
    labels={'x': 'Date', 'High': 'Price (USD)'}  # axis labels
)
fig_px.show()
````

````python
# ──────────────────────────────────────────────────────
# Method 2: Graph Objects — more control
# ──────────────────────────────────────────────────────
# go.Scatter with datetime index on x-axis
# Plotly automatically formats the datetime axis correctly
print("=== Graph Objects approach ===")
trace = go.Scatter(
    x=df_stock.index,    # DatetimeIndex on x-axis
    y=df_stock['High'],  # price values on y-axis
    name='High Price',
    line=dict(color='steelblue', width=1.5)
)

fig_go = go.Figure(data=[trace])
fig_go.update_layout(
    title='Stock High Price — Graph Objects',
    xaxis_title='Date',
    yaxis_title='Price (USD)',
    width=850, height=400
)
fig_go.show()
````

````python
# ──────────────────────────────────────────────────────
# Custom Date Range — two approaches
# ──────────────────────────────────────────────────────
from datetime import datetime

# Approach A: using datetime objects for the range
# Create three explicit datetime points
x_dates = [
    datetime(year=2016, month=1, day=1),
    datetime(year=2016, month=7, day=1),
    datetime(year=2016, month=12, day=31)
]

# Scatter trace using explicit datetime objects
trace_demo = go.Scatter(
    x=x_dates,       # datetime objects on x-axis
    y=[150, 130, 160],  # arbitrary values for demonstration
    mode='lines+markers',
    name='3 Key Dates',
    marker=dict(size=12, color='orange')
)

fig_dates = go.Figure(data=[trace_demo])
# xaxis_range: explicitly set the visible x-axis range
# Can be datetime objects (as here) or date strings (see next cell)
fig_dates.update_layout(
    title='Custom Date Range — Using datetime Objects',
    xaxis_range=[x_dates[0], x_dates[2]],  # show only between first and last dates
    width=700, height=350
)
fig_dates.show()
````

````python
# ──────────────────────────────────────────────────────
# Custom Date Range using string dates (second half of 2016)
# ──────────────────────────────────────────────────────

# Filter the stock data to only 2016 using pandas datetime-aware indexing
# df_stock.loc['2016'] selects all rows where the DatetimeIndex year is 2016
df_2016 = df_stock.loc['2016']

# Trace 1: High price in 2016 — sky blue line
trace_high_2016 = go.Scatter(
    x=df_2016.index,         # DatetimeIndex filtered to 2016
    y=df_2016['High'],        # High prices for 2016
    name='High',
    line_color='deepskyblue', # shorthand for line=dict(color='deepskyblue')
    opacity=0.8               # 80% opacity
)

# Trace 2: Low price in 2016 — grey line
trace_low_2016 = go.Scatter(
    x=df_2016.index,
    y=df_2016['Low'],
    name='Low',
    line_color='dimgray',
    opacity=0.8
)

fig_2016 = go.Figure(data=[trace_high_2016, trace_low_2016])

# xaxis_range: set using date strings 'YYYY-MM-DD'
# This shows only Jul–Dec 2016, even though we loaded all of 2016
fig_2016.update_layout(
    title='Stock High & Low — July to December 2016',
    xaxis_range=['2016-07-01', '2016-12-31'],  # date string range
    xaxis_title='Date',
    yaxis_title='Price (USD)',
    width=850, height=400
)
fig_2016.show()
````

````python
# ──────────────────────────────────────────────────────
# Time Series with Range Slider — most interactive version
# ──────────────────────────────────────────────────────
# The range slider appears below the chart and allows:
# - Drag handles to zoom into a specific time window
# - Click and drag on the slider to pan through time
# This is the most useful feature for financial time series exploration

# Full date range traces (all 2015–2017 data)
trace_high_full = go.Scatter(
    x=df_stock.index,        # full date range
    y=df_stock['High'],       # all High values
    name='High',
    line_color='deepskyblue'
)

trace_low_full = go.Scatter(
    x=df_stock.index,
    y=df_stock['Low'],
    name='Low',
    line_color='dimgray'
)

fig_slider = go.Figure(data=[trace_high_full, trace_low_full])

fig_slider.update_layout(
    title_text='Stock Price with Range Slider (2015–2017)',
    xaxis_title='Date',
    yaxis_title='Price (USD)',
    # xaxis_rangeslider_visible=True: enable the range slider below the chart
    # Users can drag this slider to zoom into any time period interactively
    xaxis_rangeslider_visible=True,
    width=900, height=500
)
fig_slider.show()
print("\nTry dragging the range slider below the chart to zoom in!")
````

## Plotly API Reference

### Figure Pattern
```python
fig = go.Figure()                   # create container
fig.add_trace(go.Scatter(...))      # add data series
fig.update_layout(title='...')      # configure appearance
fig.show()                          # render
```

### Chart Types

| Chart | Trace Class | Key Parameter |
|---|---|---|
| Line | `go.Scatter` | `mode='lines'` |
| Scatter | `go.Scatter` | `mode='markers'` |
| Bar | `go.Bar` | `barmode='group'/'stack'` |
| Table | `go.Table` | `header`, `cells` |
| Express line | `px.line()` | one-liner shortcut |

### Colour Formats

| Format | Example | Notes |
|---|---|---|
| Named | `'royalblue'` | CSS colour names |
| RGB | `'rgb(16, 112, 2)'` | No transparency |
| RGBA | `'rgba(16, 112, 2, 0.8)'` | With transparency |
| Hex | `'#1070C2'` | Hexadecimal |
